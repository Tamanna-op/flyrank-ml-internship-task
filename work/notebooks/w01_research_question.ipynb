## 2. The question: decision, action, cost of a wrong call

###Research question

Which pages should be reviewed first for refresh, expansion, protection, pruning, or monitoring, based on observable content and search signals?

###Unit of analysis

The unit of analysis is a page/content item.

###Decision

The decision is which pages should receive limited human review and attention first.

###Output

The intended output is a ranked list of pages, with a score and understandable reasons for why a page was prioritized.

###Action

A content or SEO reviewer could use the ranking to decide whether a page should be refreshed, expanded, protected, monitored, investigated further, or potentially considered for pruning or consolidation.

###Cost of a wrong recommendation

A false positive could cause the team to spend limited time reviewing or changing a page that did not need attention. A false negative could cause a potentially important page to be missed. The goal is therefore to prioritize useful candidates for human review rather than automatically making the final content decision.

## 3. Quick look at the data (2-3 real numbers)
The starter dataset contains 30,000 page records, of which 16,262 (54.2%) are labeled as declining using the starter label. The median number of days since the last update is 20 days.

These numbers make the problem worth investigating because the dataset contains a large number of pages, making manual review of every page impractical. It also contains page-level search and content signals such as impressions, average position, CTR, content age, and days since the last update that may provide useful information for prioritizing pages.

I also found that word_count has 7,699 missing values, so any future use of this feature will need appropriate handling of missing data.

These observations support investigating the lane, but they do not show that refreshing a page will cause its performance to improve.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Load the starter dataset and calculate a few real numbers.

import pandas as pd

df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Create the starter declining-page label.
df["is_declining_label"] = (
    df["trend_direction"].str.lower().eq("down").astype(int)
)

print("Number of pages:", len(df))
print("Declining pages:", df["is_declining_label"].sum())
print(
    "Declining rate:",
    f"{df['is_declining_label'].mean() * 100:.1f}%"
)

print(
    "Median days since last update:",
    df["days_since_last_update"].median()
)
```
